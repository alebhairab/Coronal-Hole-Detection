```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# ---- 1. Load data ----
INPUT_FILE = "/home/bhairab/Downloads/whole coronal hole.xlsx"
OUTPUT_DIR = "/home/bhairab/Documents/project/plot"
os.makedirs(OUTPUT_DIR, exist_ok=True)

df = pd.read_excel(INPUT_FILE, sheet_name="all")

# Keep only the value columns; drop empty rows (Date column intentionally ignored)
df = df[["AIA (DN)", "HMI (Mx)", "Area (meter^2)"]].dropna()
df = df.reset_index(drop=True)

# ---- 2. Manually build DateTime column ----
START_DATE = pd.Timestamp("2025-09-08")
HOURS = [3, 6, 9, 12, 15, 18, 21, 23]  # 8 observations per day

datetimes = []
day_offset = 0
hour_idx = 0
for _ in range(len(df)):
    dt = START_DATE + pd.Timedelta(days=day_offset, hours=HOURS[hour_idx])
    datetimes.append(dt)
    hour_idx += 1
    if hour_idx == len(HOURS):
        hour_idx = 0
        day_offset += 1

df["DateTime"] = datetimes

# ---- 3. Global style to
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "DejaVu Serif"],
    "font.size": 16,
    "axes.linewidth": 1.5,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "xtick.major.size": 8,
    "ytick.major.size": 8,
    "xtick.minor.size": 4,
    "ytick.minor.size": 4,
    "xtick.major.width": 1.5,
    "ytick.major.width": 1.5,
})


def style_axes(ax):
    ax.xaxis.set_major_locator(mdates.DayLocator())
    ax.xaxis.set_minor_locator(mdates.HourLocator(interval=3))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%-d %b\n%Y"))
    ax.tick_params(which="both", top=True, right=True)
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)


# ---- 4. Plot AIA (DN) ----
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df["DateTime"], df["AIA (DN)"], "-o", color="black",
        markersize=6, linewidth=1.5, markerfacecolor="black")
ax.set_xlabel("Date")
ax.set_ylabel("AIA (DN)")
style_axes(ax)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "aia_plot.png"), dpi=300)
plt.close()

# ---- 5. Plot HMI (Mx) ----
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df["DateTime"], df["HMI (Mx)"], "-o", color="black",
        markersize=6, linewidth=1.5, markerfacecolor="black")
ax.set_xlabel("Date")
ax.set_ylabel("HMI (Mx)")
style_axes(ax)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "hmi_plot.png"), dpi=300)
plt.close()

# ---- 6. Plot Area (m^2) ----
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df["DateTime"], df["Area (meter^2)"], "-o", color="black",
        markersize=6, linewidth=1.5, markerfacecolor="black")
ax.set_xlabel("Date")
ax.set_ylabel(r"Area of Coronal Hole (m$^2$)")
style_axes(ax)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "area_plot.png"), dpi=300)
plt.close()

print(f"Saved plots to {OUTPUT_DIR}")
```
